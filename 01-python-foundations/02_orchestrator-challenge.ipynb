```python
import asyncio
import random


async def procesar_biometricos(usuario: str):
    # Simula latencia de red de un Agente de IA
    await asyncio.sleep(random.uniform(0.5, 2.0))
    ritmo = random.randint(60, 120)
    
    if ritmo > 100:
        print(f"!!! ALERTA CRÍTICA para {usuario}: {ritmo} bpm !!!")
        
    return f"{usuario}: {ritmo} bpm"


async def main():
    usuarios = ["Rodrigo", "Daniela", "Paciente_01", "Paciente_02", "Paciente_03"]
    
    print("Iniciando procesamiento de biométricos...")
    resultados = await asyncio.gather(*(procesar_biometricos(u) for u in usuarios))
    
    print("\nResultados finales:")
    for res in resultados:
        print(res)


# Para ejecutar en notebook
await main()

# Para ejecutar en .py
"""
if __name__ == "__main__":
    asyncio.run(main())
"""
```
